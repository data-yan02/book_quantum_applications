```python
#Import the required libraries
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from qiskit.quantum_info import Statevector
import numpy as np
import matplotlib
```

```python
#Use AerSimulator for quantum simulation
backend = AerSimulator()
shots = 1024
style = {'backgroundcolor': 'white'} # Style of the circuits
qreg1 = QuantumRegister(2) # The 2 qubits quantum register
register1 = ClassicalRegister(1)
qc = QuantumCircuit(qreg1, register1)
```

```python
qc.x(1)
qc.barrier()
qc.draw(output='mpl',style=style)
```

```python
qc.h(0)
qc.h(1)
qc.barrier()
qc.draw(output='mpl',style=style)
```

```python
qc.cx(0,1)
qc.barrier()
qc.h(0)
qc.draw(output='mpl',style=style)
```

```python
qc.measure(qreg1[0],register1)
qc.draw(output='mpl',style=style)
```

```python
results = backend.run(transpile(qc, backend), shots=shots).result()
answer = results.get_counts()
plot_histogram(answer)
```

```python
Qreg = QuantumRegister(2)
Creg = ClassicalRegister(1)
qcirc = QuantumCircuit(Qreg, Creg)

qcirc.x(1)

qcirc.h(0)
qcirc.h(1)
qcirc.barrier()

qcirc.x(1)
qcirc.barrier()

qcirc.h(0)
qcirc.barrier()

qcirc.measure(Qreg[0],Creg)
qcirc.draw(output='mpl',style=style)
```
